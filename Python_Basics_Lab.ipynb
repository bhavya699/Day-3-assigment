## Random Numbers and String Reversal

### Program A: Create a list of random numbers and print even numbers

````
import random
random_numbers=[random.randint(1,100) for _ in range(15)]
print("Random Numbers:",random_numbers)
print("Even Numbers:")
for n in random_numbers:
    if n%2==0:
        print(n,end=" ")
````

**Sample Output**
```
Random Numbers: [45,12,67,84,29,56,91,38,73,44,18,97,20,63,88]
Even Numbers:
12 84 56 38 44 18 20 88
```

### Program B: Reverse a String

````
text="Python Programming"
print("Original:",text)
print("Reversed:",text[::-1])
````

**Sample Output**
```
Original: Python Programming
Reversed: gnimmargorP nohtyP
```

### Reverse the Order of Words

````
text="Python is very easy"
print("Original:",text)
print("Words Reversed:"," ".join(text.split()[::-1]))
````

**Sample Output**
```
Original: Python is very easy
Words Reversed: easy very is Python
```
